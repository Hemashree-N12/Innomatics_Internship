## *Problem Statement 1: Employee Performance Bonus Eligibility*

```python
#Display all the employee details with their performance scores in a dictionary
employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
}
#Find the highest performance score value
highest = max(employees.values())
#Find the employee with the highest score
best_employee = []
for name,score in employees.items():
    if (score==highest):
        best_employee.append(name)
#Display 
print("Top Performers Eligible for Bonus:",best_employee,"(score:",highest,")")
```

## *Problem Statement 2: Search Query Keyword Analysis*

```python
queries = "Buy mobile phone buy phone online"
#convert to lowercase
queries = queries.lower()
#remove punctuation
punctuation = ".,!?;:" #define what punctuations to remove
for i in punctuation:
    queries = queries.replace(i,"")
#split the words to get count
words = queries.split()
#count the frequency
count = {}
for word in words:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1
#Display only keywords occuring more than once
multiple_counts = {}
for word in count:
    if(count[word]>1):
        multiple_counts[word] = count[word]
print(multiple_counts)
```

## *Problem Statement 3: Sensor Data Validation*

```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
#store valid readings
valid = []
#loop through readings
for i in range(len(sensor_readings)):
    value = sensor_readings[i] #fetch the index value
    if value%2 == 0: #chcek for even values
        valid.append((i,value)) #stores the data as  (hour_index, reading_value) pairs.
print("Valid Sensor Readings (Hour, Value):")
print(valid)
```

## *Problem Statement 4: Email Domain Usage Analysis*

```python
#Given data
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
#Check the domains
domains = []
for email in emails:
    domain = email.split("@")[1] #takes the value after @
    domains.append(domain)
#Count the frequency
count = { }
for domain in domains:
    if domain in count:
        count[domain] += 1
    else:
        count[domain] = 1
#To calculate percentage, we check th total no of emails
total = len(emails)
percentage = { }
for domain in domains:
    percent = (count[domain]/total) * 100
    percentage[domain] = percent
#Display
for domain, percent in percentage.items():
    print(f"{domain}:{percent}%")
```

## *Problem Statement 5: Sales Spike Detection*

```python
#Daily sales data
sales = [1200, 1500, 900, 2200, 1400, 3000]
#Calculate average sales 
average = sum(sales)/len(sales)
#to check the data for 30% more than average limit set threshold 
threshold = average * 1.3
#check for days meeting the threshold limit
#initialize day value as 1 to counting from 1st day
day = 1
for i in sales:
    if i > threshold:
        print("Day",day,":",i)
    day += 1 #to loop through the days
```

## *Problem Statement 6: Duplicate User ID Detection*

```python
#User ID'S are
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
#count frequency
count = { }
for ids in user_ids:
    if ids in count:
        count[ids] += 1
    else:
        count[ids] = 1
#to filter duplicate values
duplicate  = { }
for ids in count:
    if count[ids] > 1:
        duplicate[ids] = count[ids]
#Display
for ids, count in duplicate.items():
    print(ids, "→", count, "times")
```
