## **Problem 1: Social Media – Post Engagement Analyzer**

```python
#define a function
def analyze(likes):
    total_likes = 0
    #loop through each like and add it
    for like in likes:
        total_likes += like
    #Check the status
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    print("Total Likes:", total_likes)
    print("Post Status:", status)
#Use sample data from a few posts
analyze([250,500,300])
```

## **Problem 3: Agriculture – Rainfall Adequacy Checker**

```python
def check_rainfall(rainfall, level_required):
    #calculate the total amount of seasonal rainfall
    total = sum(rainfall)
    # Find the average rainfall in the region
    average = total / len(rainfall)
    if average >= level_required:
        status = "Adequate Rainfall" 
    else:
        status = "Inadequate Rainfall"
    print("Average Rainfall:", average)
    print("Rainfall Status:", status)
#Sample data
check_rainfall([60, 80, 76], 70)
```

## **Problem 4: Social Media – Duplicate Account Detection**

```python
def duplicate(usernames):
    #We use set to remove duplicate username, so if the len of set is not equal the duplicate values exist
    if len(usernames) != len(set(usernames)): 
        flag = "Yes"
    else:
        flag = "No"
    print("Duplicate Account Found: ",flag)
#Sample data
duplicate(["abcde","ad123","Lola3","abcde"])
```

## **Problem 6: Agriculture – Premium Crop Price Filter**

```python
def crop_filter(price):
    #Check the crop price category
    premium = [] #declare as list to store the values
    status_list = []
    for p in price:
        if p > 2000:
            premium.append(p) #we use append to add to the list
            status_list.append("Premium Crop") 
        else:
            status_list.append("Regular Crop")
    print("Premium Crops:",premium)
    print("Crop Status:",status_list)
#Check for sample data
crop_filter([1500,2100,3500,1800])
```

## **Problem 7: System Monitoring – Application Health Checker**

```python
def system_checker(errors):
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    print("Error Count:",errors)
    print("System Status:",status)
#check for sample data
system_checker(7)
```

## **Problem 9: E-Learning – Student Attendance Eligibility System**

```python
def eligibility_checker(attendance): 
    #check for number of days persent out of total days and check for percentage
    total = len(attendance)
    present = 0
    for day in attendance:
        if day == "P":
            present += 1
        percentage = ((present/total)*100)
    if percentage >= 75:
        status = "Approved"
    else:
        status = "Rejected"
    print("Attendance Percentage:",percentage)
    print("Exam Eligibility:",status)
#check for sample data
eligibility_checker(["P","A","P","P","P"])
```

## **Problem 10: Smart Electricity Bill Analyzer**

```python
def electricity_bill(units):
    if units <= 100:
        bill = units * 3 #First 100 units → ₹3 per unit
    elif units <= 200:
        bill = (100 * 3) + (units - 100) * 5 #Next 100 units (101–200) → ₹5 per unit 
    else:
        bill = (100 * 3) + (100 * 5) + (units - 200) * 7 #Above 200 units → ₹7 per unit
    #After calculating the bill check the usage of the electricity according to the bill
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    print("Total Bill:", bill)
    print("Usage Status:", usage)
#Sample alue
electricity_bill(300)
```
