## **Problem Statement 1:Smart Parking Lot Management System**

```python
def parking_management(capacity, logs):
    parked = 0 #initialize parked vehicles to 0
    peak = 0 #initialize usage at peak to 0
    #check entry and exit logs
    for log in logs:
        if log == "IN":
            parked += 1
        else:
            if log == "OUT" and parked > 0:
                parked -= 1
        #Track the usage at peak 
        if parked > peak:
            peak = parked
        #Send alerts if maximum capacity is reached
        if parked > capacity:
            print("Alert!! Capacity is full, Parking space has exceeded.")
    #Check the parking status
    if parked < capacity:
         status = "Available"
    else:
        status = "Full"
    #Display the details    
    print("Currently Parked Vehicles:", parked)
    print("Parking Status:", status)
# Sample Input
capacity = 50
parking_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
parking_management(capacity, parking_logs)
```

## **Problem Statement 2:Online Food Delivery Time Estimator**

```python
def delivery(distance,weather,traffic):
    #set base time as 4 minutes per km
    time = distance*4
    delay = 0
    
    #Traffic Delays
    if traffic == "High":
        delay += 15
    else:
        if traffic == "Medium":
            delay += 10
        else:
            delay += 0

    #Weather Delays
    if weather in ["Rainy","Windy","Foggy"]:
        delay += 8
    else:
        if weather in ["Snowy","Stromy"]:
            delay += 20
        else:
            delay += 0

    #define the total time/eta
    total = time+delay
    print("Estimated Delivery Time:",total, "minutes")

#For sample data
delivery(8,"Rainy","High")
```

## **Problem Statement 4:Cloud Server Load Classification System**

```python
def server_load(readings):
    average = sum(readings) / len(readings)
    #Check classification
    if average < 50:
        status = "Normal"
    elif average <= 80:
        status = "Warning"
    else:
        status = "Critical"
    #Display the details
    print("Average CPU Load:", int(average), "%")
    print("Server Status:", status)
# Sample Input
server_load([45, 60, 70, 85, 90])
```

## **Problem Statement 5:Smart Classroom Resource Usage Monitor**

```python
def class_usage(resources):
    overused = [] #create an empty list
    for item, hours in resources.items(): #check for items and hours used with the help of a dictionary
        #set threshold limit
        if hours >= 7:
            overused.append(item) #adds the items having threshold value greater than or equal to 7 to the list
    alert = "Yes" if overused else "No"
    usage = ", ".join(overused) if overused else "None"
    #Display the details
    print("Overused resources:", usage)
    print("Energy Alert:", alert)
# Sample Data
use = {"Projector": 6, "AC": 9, "Lights": 4}
class_usage(use)
```

## **Problem Statement 6: Online Event Registration Capacity Controller**

```python
def event(capacity, registration):
    confirmed = 0
    waitlist = 0
    if registration > capacity:
        confirmed = capacity
        waitlist = registration - capacity
        status = "Closed"
    else:
        confirmed = registration
        waitlist = 0
        status = "Open"
    #Display the statements
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)
    print("Registration Status:", status)
# Sample Data
event(100, 105)
```
