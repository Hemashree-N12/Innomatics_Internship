## **Problem 2: Inventory Reorder Alert System**

```python
#define a function
def inventory(products):
    #loop through each product
    for item, stock in products.items():
        if stock < 15: #Alert value
            print(item,"-> Reorder Alert")
        else:
            print(item,"-> Stock OK")
            
#sample dictionary with products
products = {
    "Paint" : 14,
    "Brushes" : 28,
    "Sketchpad" : 30,
    "Pencils" : 10
}

#Call the function
inventory(products)
```

## **Problem 5: Employee Attendance Eligibility Checker**


```python
#define a function
def employee(attendance):
    total_days = len(attendance)
    present = 0
    
    #loop to count number of days present
    for day in attendance:
        if day == "P":
            present += 1
    percentage = ((present/total_days)*100) #calculate percentage
    
    #check eligibility
    if percentage >= 75: #Minimum percentage
        print("Eligible") 
    else:
        print("Not Eligible")
    return percentage, total_days, present
    
#attendance 
Attendance = [ "P", "A", "P", "A", "P", "P"]
Percentage, Total_days, Present = employee(Attendance) #Call the function

#display
print("Total days of work:",Total_days)
print("Number of days present:",Present)
print("Percentage of days Present:",Percentage)
```

## **Problem 6: Password Strength Checker**

```python
#define a function
def check(password):
    # conditions for a valid password
    if len(password) < 8:
        return "Password too short (minimum 8 characters)."

    #it should have minimum 1 digit
    has_digit = False
    for ch in password:
        if ch.isdigit():
            has_digit = True
            break
    if not has_digit:
        return "Password must have at least one digit"

    #For special character @, #, $
    has_special = False
    for ch in password:
        if ch in "@#$":
            has_special = True
            break
    if not has_special:
            return "Password must have at least one special character (@, #, $)"
    return "Password is valid"

# retry loop
while True:
    pwd = input("Enter password: ") #take input for password
    result = check(pwd)
    print(result)
    if result == "Password is valid":
        break
    else:
        print("Please try again...\n")
```
