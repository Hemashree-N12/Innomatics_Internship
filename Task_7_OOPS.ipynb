## **Problem 1: Smart Light Controller**

```python
#create a class 
class Light:
    #create a method
    def __init__(self,name):
        self.name = name #store the light's name
        self.status = "OFF" #default status is off
    #create method for controlling the light (ON and OFF)
    def control(self,action):
        self.status = action #update the status
        print("Light Name:",self.name)
        print("Action:",self.status)
#create an object
light = Light("Bedroom Light")
#call the method
light.control("ON")
```

## **Problem 2: Employee ID Card System**

```python
class Employee:
    def __init__(self,name,id,department):
        self.name = name
        self.id = id
        self.department = department
    #define the ID Card method
    def ID_Card(self):
        print("Employee ID Card")
        print("Name:",self.name)
        print("ID:",self.id)
        print("Department:",self.department)
#create an object
employee = Employee("Rahul","EMP102","IT")
#method calling
employee.ID_Card()
```

## **Problem 3: Mobile Contact Record**

```python
class Contact:
    def __init__(self,name,number):
        self.name = name
        self.number = number
    #define method to display the details
    def display_contact(self):
        print("Contact Saved")
        print("Name:",self.name)
        print("Phone Number:",self.number)
#create an object
contact = Contact("Anitha","9876543210")
#method calling
contact.display_contact()
```

## **Problem 4: Product Price Tag Generator**

```python
class Product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    #define method to display the details
    def display_price_tag(self):
        print("Product Name:",self.name)
        print("Price: ₹"+self.price)
#create an object
product = Product("Headphones","2499")
#method calling
product.display_price_tag()
```

## **Problem 5: Movie Rating Display System**

```python
class Movie:
    def __init__(self,name,rating):
        self.name = name
        self.rating = rating
    #define method to display the details
    def display_rating(self):
        print("Movie Name:",self.name)
        print("Rating:",self.rating+"/5")
#create an object
movie = Movie("Inception","4.8")
#method calling
movie.display_rating()
```

## **Problem 6: Delivery Address Manager**

```python
class Delivery:
    def __init__(self,name,address):
        self.name = name
        self.address = address
    #define method to display the details
    def display_customer(self):
        print("Delivery Details")
        print("Customer Name:",self.name)
        print("Address:",self.address)
#create an object
delivery = Delivery("Suman","Hyderabad")
#method calling
delivery.display_customer()
```
